# AREA B — REVENUE AND ARPU (B1–B9)

## B1 — Average stated ARPU

Calculate the average stated_arpu across subscribers. Also show the number of subscribers included.

```python
q("""
SELECT ROUND(AVG(stated_arpu), 2) AS avg_arpu, -- average stated plan price
       COUNT(*) AS subs                          -- number of subscribers
FROM customer_features;
""")
```

## B2 — Average actual ARPU and monthly total

Calculate the average actual_arpu_6m and the total monthly revenue represented by this measure.

```python
q("""
SELECT ROUND(AVG(actual_arpu_6m), 2) AS avg_actual, -- average money received
       ROUND(SUM(actual_arpu_6m), 0) AS monthly_total -- total monthly revenue
FROM customer_features;
""")
```

## B3 — Difference between stated and actual ARPU

Compare stated_arpu with actual_arpu_6m. Calculate the average gap and percentage difference. State the denominator used.

```python
q("""
SELECT ROUND(AVG(stated_arpu), 2) AS stated,       -- average stated amount
       ROUND(AVG(actual_arpu_6m), 2) AS actual,    -- average received amount
       ROUND(AVG(arpu_gap), 2) AS gap,             -- average difference
       ROUND(100.0 * (SUM(stated_arpu) - SUM(actual_arpu_6m))
             / NULLIF(SUM(actual_arpu_6m), 0), 2) AS over_pct -- denominator: actual revenue
FROM customer_features;
""")
```

# AREA E — USAGE INTENSITY AND DATA ALLOWANCE (E1–E8)

## E1 — Four quartiles of daily data use

Exclude subscribers with no usage. Divide subscribers into four equal groups using NTILE(4) ordered by mb_per_active_day. Show group size and minimum and maximum daily use.

```python
q("""
WITH n AS (
    SELECT mb_per_active_day AS m,               -- daily data measure
           NTILE(4) OVER (ORDER BY mb_per_active_day) AS q -- four equal groups
    FROM customer_features
    WHERE usage_days > 0                         -- exclude no-usage subscribers
)
SELECT q, COUNT(*) AS subs,                      -- quartile and group size
       ROUND(MIN(m), 0) AS mn,                   -- minimum daily use
       ROUND(MAX(m), 0) AS mx                    -- maximum daily use
FROM n
GROUP BY q
ORDER BY q;
""")
```

## E2 — Churn by daily data-use quartile

Using the same four daily-data groups, calculate churn percentage in each group.

```python
q("""
WITH n AS (
    SELECT is_churned,                           -- carry churn flag
           NTILE(4) OVER (ORDER BY mb_per_active_day) AS q -- daily data quartile
    FROM customer_features
    WHERE usage_days > 0
)
SELECT q, COUNT(*) AS subs,                      -- quartile size
       ROUND(100.0 * AVG(is_churned), 2) AS churn -- churn percentage
FROM n
GROUP BY q
ORDER BY q;
""")
```

## E3 — Voice and data use by connection type and device

Compare voice minutes per active day and data MB per active day for each connection_type and device_type combination.

```python
q("""
SELECT connection_type, device_type,            -- subscriber segment
       COUNT(*) AS subs,                         -- subscribers in segment
       ROUND(AVG(total_voice_min * 1.0 /
                 NULLIF(usage_days, 0)), 1) AS voice_per_day, -- voice per active day
       ROUND(AVG(mb_per_active_day), 0) AS mb_per_day -- data per active day
FROM customer_features
WHERE usage_days > 0
GROUP BY connection_type, device_type
ORDER BY voice_per_day DESC;
""")
```

## E4 — Monthly allowance compared with actual monthly use by plan

For each plan, compare monthly data allowance (monthly_gb) with average actual monthly data use in GB.

```python
q("""
SELECT plan_name, monthly_gb,                 -- plan and monthly allowance
       ROUND(AVG(total_data_mb) / 1024.0 / 6, 2) AS actual_gb_per_month
                                                  -- six-month data converted to monthly GB
FROM customer_features
WHERE usage_days > 0
GROUP BY plan_name, monthly_gb
ORDER BY monthly_gb;
""")
```

## E5 — Overall allowance usage and subscribers below half allowance

Calculate average monthly allowance, average monthly usage, average percentage of allowance used, and number of subscribers using less than half their allowance.

```python
q("""
SELECT ROUND(AVG(monthly_gb), 1) AS allowance_gb, -- average monthly allowance
       ROUND(AVG(total_data_mb / 1024.0 / 6), 2) AS used_gb, -- average monthly use
       ROUND(100.0 * AVG((total_data_mb / 1024.0 / 6) /
                         NULLIF(monthly_gb, 0)), 1) AS pct_of_allowance,
                                                  -- average subscriber-level share
       SUM((total_data_mb / 1024.0 / 6) < monthly_gb * 0.5) AS under_half
                                                  -- count below 50 percent
FROM customer_features
WHERE usage_days > 0;
""")
```

## E6 — Spend, data use and churn by monetary score

Group subscribers into five monetary score groups and compare six-month spend, data per active day and churn.

```python
q("""
WITH scored AS (
    SELECT
        NTILE(5) OVER (ORDER BY recharge_total_6m) AS m_score,
        recharge_total_6m,
        mb_per_active_day,
        is_churned
    FROM customer_features
)
SELECT
    m_score,
    COUNT(*) AS subs,
    ROUND(AVG(recharge_total_6m), 0) AS spend_6m,
    ROUND(AVG(mb_per_active_day), 0) AS mb_per_day,
    ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM scored
GROUP BY m_score
ORDER BY m_score;
""")
```

## E7 — Churn by usage-days quartile

Divide subscribers into four equal groups by usage_days. Show each group's usage-day range and churn percentage.

```python
q("""
WITH n AS (
    SELECT usage_days, is_churned,               -- carry days and churn flag
           NTILE(4) OVER (ORDER BY usage_days) AS q -- four equal groups
    FROM customer_features
)
SELECT q, COUNT(*) AS subs,                      -- quartile size
       MIN(usage_days) AS mn,                     -- minimum active days
       MAX(usage_days) AS mx,                     -- maximum active days
       ROUND(100.0 * AVG(is_churned), 2) AS churn -- churn percentage
FROM n
GROUP BY q
ORDER BY q;
""")
```

## E8 — SMS per active day and subscribers who never sent SMS

Calculate average SMS sent per active day, count subscribers who never sent an SMS, and show the population included.

```python
q("""
SELECT ROUND(AVG(total_sms * 1.0 /
                 NULLIF(usage_days, 0)), 3) AS sms_per_active_day,
                                                  -- average SMS per active day
       SUM(total_sms = 0) AS never_sent_one,      -- subscribers with zero SMS
       COUNT(*) AS subs                           -- population included
FROM customer_features
WHERE usage_days > 0;
""")
```

# AREA G — COMPLAINTS

## G1 — Most common complaint categories

What do subscribers complain about most?

```python
q("""
SELECT category,
       COUNT(*) AS tickets,
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1) AS pct
FROM complaints
GROUP BY category
ORDER BY tickets DESC;
""")
```

## G2 — Resolution rate by complaint channel

Does the complaint channel change how often a ticket is resolved?

```python
q("""
SELECT channel,
       COUNT(*) AS tickets,
       ROUND(100.0 * SUM(is_resolved) / COUNT(*), 2) AS pct_resolved
FROM complaints
GROUP BY channel
ORDER BY tickets DESC;
""")
```

## G3 — Open and resolved tickets

How many tickets are still open?

```python
q("""
SELECT is_resolved,
       COUNT(*) AS tickets,
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS pct
FROM complaints
GROUP BY is_resolved;
""")
```

## G4 — Resolution rate and time by category

Which complaint category is resolved least often, and most slowly?

```python
q("""
SELECT category,
       COUNT(*) AS tickets,
       ROUND(100.0 * SUM(is_resolved) / COUNT(*), 2) AS pct_resolved,
       ROUND(AVG(CASE WHEN is_resolved = 1 THEN resolution_days END), 2) AS avg_days
FROM complaints
GROUP BY category
ORDER BY pct_resolved;
""")
```

## G5 — Time to resolve a ticket

How many days does a resolved ticket take to close?

```python
q("""
SELECT ROUND(AVG(resolution_days), 2) AS avg_days,
       MIN(resolution_days) AS mn,
       MAX(resolution_days) AS mx,
       COUNT(*) AS closed
FROM complaints
WHERE is_resolved = 1;
""")
```

## G6 — Churn among subscribers who complain

```python
q("""
SELECT
    SUM(complaint_count = 0) AS never_,
    SUM(complaint_count > 0) AS did,
    ROUND(100.0 * AVG(CASE WHEN complaint_count = 0 THEN is_churned END), 2) AS churn_never,
    ROUND(100.0 * AVG(CASE WHEN complaint_count > 0 THEN is_churned END), 2) AS churn_did
FROM customer_features;
""")
```

## G7 — Subscriber distribution by complaint count

## I1 — Overall subscriber churn

How many subscribers have we lost, and what is the churn rate?

```python
q("""
SELECT COUNT(*) AS subs,
       SUM(is_churned) AS left_,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features;
""")
```

## I2 — Churn by connection type

What is the churn rate for prepaid and for postpaid subscribers?

```python
q("""
SELECT connection_type,
       COUNT(*) AS subs,
       SUM(is_churned) AS ch,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY connection_type;
""")
```

## I3 — Plans with highest churn

Which six plans have the highest churn rate?

```python
q("""
SELECT plan_name,
       plan_type,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY plan_name, plan_type
ORDER BY churn DESC
LIMIT 6;
""")
```

## I4 — Churn by circle and region

Which circles have the highest and the lowest churn rate?

```python
q("""
SELECT circle_name,
       region,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY circle_name, region
ORDER BY churn DESC;
""")
```

## I5 — Churn by device type

Do smartphone users leave at a different rate from feature-phone users?

```python
q("""
SELECT device_type,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY device_type;
""")
```

## I6 — Churn by acquisition channel

Does churn differ by acquisition channel?

```python
q("""
SELECT acquisition_channel,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY acquisition_channel
ORDER BY churn DESC;
""")
```

## I7 — Churn by age band

Does churn differ by age band?

```python
q("""
SELECT age_band,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY age_band
ORDER BY age_band;
""")
```

## I8 — Churn spread across subscriber attributes

Which subscriber attribute separates leavers from stayers the most?

```python
q("""
SELECT 'circle' AS candidate, ROUND(MAX(churn_rate) - MIN(churn_rate), 2) AS spread
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY circle_name
)
UNION ALL
SELECT 'plan', ROUND(MAX(churn_rate) - MIN(churn_rate), 2)
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY plan_name
)
UNION ALL
SELECT 'connection_type', ROUND(MAX(churn_rate) - MIN(churn_rate), 2)
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY connection_type
)
UNION ALL
SELECT 'device_type', ROUND(MAX(churn_rate) - MIN(churn_rate), 2)
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY device_type
)
UNION ALL
SELECT 'age_band', ROUND(MAX(churn_rate) - MIN(churn_rate), 2)
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY age_band
)
UNION ALL
SELECT 'acquisition_channel', ROUND(MAX(churn_rate) - MIN(churn_rate), 2)
FROM (
    SELECT 100.0 * AVG(is_churned) AS churn_rate
    FROM customer_features GROUP BY acquisition_channel
)
ORDER BY spread DESC;
""")
```

# AREA J — CHURN: WHY THEY LEAVE

## J1 — Churn by complaint resolution status

Does an open complaint make a subscriber more likely to leave?

```python
q("""
SELECT
    CASE
        WHEN complaint_count = 0 THEN '1 never complained'
        WHEN unresolved_count = 0 THEN '2 all resolved'
        ELSE '3 has open ticket'
    END AS grp,
    COUNT(*) AS subs,
    ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY grp
ORDER BY grp;
""")
```

## J2 — Churn by number of open tickets

Does churn rise with each extra open ticket?

```python
q("""
SELECT unresolved_count,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY unresolved_count
ORDER BY unresolved_count;
""")
```

## J3 — Churn by complaint category

Does what a subscriber complained about change their churn rate?

```python
q("""
WITH subscriber_categories AS (
    SELECT DISTINCT customer_id, category
    FROM complaints
)
SELECT x.category,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(c.is_churned), 2) AS churn
FROM subscriber_categories x
JOIN customer_features c
  ON c.customer_id = x.customer_id
GROUP BY x.category
ORDER BY churn DESC;
""")
```

## J4 — Churn by complaint resolution time

Do slower resolutions lead to more churn?

```python
q("""
SELECT
    CASE
        WHEN avg_resolution_days IS NULL THEN 'none closed'
        WHEN avg_resolution_days <= 4 THEN '1-4 days'
        WHEN avg_resolution_days <= 8 THEN '5-8 days'
        ELSE '9-14 days'
    END AS band,
    COUNT(*) AS subs,
    ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
WHERE complaint_count > 0
GROUP BY band
ORDER BY band;
""")
```

## J5 — Usage differences between leavers and stayers

How did leavers' usage differ from stayers'?

```python
q("""
SELECT is_churned,
       COUNT(*) AS subs,
       ROUND(AVG(usage_days), 1) AS active_days,
       ROUND(AVG(mb_per_active_day), 0) AS mb_per_day,
       ROUND(AVG(recency_days), 1) AS days_quiet
FROM customer_features
WHERE usage_days > 0
GROUP BY is_churned;
""")
```

## J6 — Payment differences between leavers and stayers

How did leavers' payments differ from stayers'?

```python
q("""
SELECT is_churned,
       COUNT(*) AS subs,
       ROUND(AVG(recharge_count), 2) AS payments,
       ROUND(AVG(actual_arpu_6m), 2) AS arpu
FROM customer_features
GROUP BY is_churned;
""")
```

## J7 — Monthly data use for leavers and stayers

How did leavers' monthly data use compare with stayers', month by month?

```python
q("""
SELECT m.month,
       ROUND(AVG(CASE WHEN c.is_churned = 1 THEN m.total_mb / 1024.0 END), 2) AS gb_leavers,
       ROUND(AVG(CASE WHEN c.is_churned = 0 THEN m.total_mb / 1024.0 END), 2) AS gb_stayers
FROM v_usage_month m
JOIN customer_features c
  ON c.customer_id = m.customer_id
GROUP BY m.month
ORDER BY m.month;
""")
```

## J8 — Churn gaps across spend, tenure, connection and open tickets

Of spend, tenure, connection type and open tickets, which moves churn most?

```python
q("""
WITH scored AS (
    SELECT *,
           NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score
    FROM customer_features
),
gaps AS (
    SELECT 'spend: lowest vs highest' AS factor,
           100.0 * AVG(CASE WHEN m_score = 4 THEN is_churned END)
         - 100.0 * AVG(CASE WHEN m_score = 1 THEN is_churned END) AS churn_gap
    FROM scored
    UNION ALL
    SELECT 'tenure: shortest vs longest',
           100.0 * AVG(CASE WHEN tenure_years = (SELECT MIN(tenure_years) FROM scored) THEN is_churned END)
         - 100.0 * AVG(CASE WHEN tenure_years = (SELECT MAX(tenure_years) FROM scored) THEN is_churned END)
    FROM scored
    UNION ALL
    SELECT 'connection: postpaid vs prepaid',
           100.0 * AVG(CASE WHEN connection_type = 'Postpaid' THEN is_churned END)
         - 100.0 * AVG(CASE WHEN connection_type = 'Prepaid' THEN is_churned END)
    FROM scored
    UNION ALL
    SELECT 'open tickets: any vs none',
           100.0 * AVG(CASE WHEN unresolved_count > 0 THEN is_churned END)
         - 100.0 * AVG(CASE WHEN unresolved_count = 0 THEN is_churned END)
    FROM scored
)
SELECT factor, ROUND(churn_gap, 2) AS churn_gap_pp
FROM gaps
ORDER BY ABS(churn_gap) DESC;
""")
```

# AREA K — RECENCY, FREQUENCY AND MONETARY

## K1 — Churn by recency quartile

What is the churn rate in each recency quartile?

```python
q("""
WITH scored AS (
    SELECT NTILE(4) OVER (ORDER BY recency_days DESC) AS r_score,
           is_churned
    FROM customer_features
)
SELECT r_score,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM scored
GROUP BY r_score
ORDER BY r_score;
""")
```

## K2 — Churn by frequency quartile

What is the churn rate in each frequency quartile?

```python
q("""
WITH scored AS (
    SELECT NTILE(4) OVER (ORDER BY recharge_count) AS f_score,
           recharge_count,
           is_churned
    FROM customer_features
)
SELECT f_score,
       COUNT(*) AS subs,
       ROUND(AVG(recharge_count), 2) AS payments,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM scored
GROUP BY f_score
ORDER BY f_score;
""")
```

## K3 — Spend and churn by monetary quartile

What are the spend and the churn rate in each spend quartile?

```python
q("""
WITH scored AS (
    SELECT NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score,
           recharge_total_6m,
           is_churned
    FROM customer_features
)
SELECT m_score,
       COUNT(*) AS subs,
       ROUND(AVG(recharge_total_6m), 0) AS spend,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM scored
GROUP BY m_score
ORDER BY m_score;
""")
```

## K5 — Four frequency and monetary segments

How big is each of four payment segments, and how often does each leave?

```python
q("""
WITH scored AS (
    SELECT *,
           NTILE(4) OVER (ORDER BY recharge_count) AS f_score,
           NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score
    FROM customer_features
),
segments AS (
    SELECT *,
        CASE
            WHEN f_score >= 3 AND m_score >= 3 THEN 'Best'
            WHEN f_score >= 3 AND m_score < 3 THEN 'Frequent low value'
            WHEN f_score < 3 AND m_score >= 3 THEN 'Big but rare'
            ELSE 'At risk'
        END AS seg
    FROM scored
)
SELECT seg,
       COUNT(*) AS subs,
       ROUND(AVG(recharge_total_6m), 0) AS spend,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM segments
GROUP BY seg
ORDER BY churn;
""")
```

## K6 — Monthly revenue share by segment

What share of monthly revenue does each segment carry?

```python
q("""
WITH scored AS (
    SELECT *,
           NTILE(4) OVER (ORDER BY recharge_count) AS f_score,
           NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score
    FROM customer_features
),
segments AS (
    SELECT *,
        CASE
            WHEN f_score >= 3 AND m_score >= 3 THEN 'Best'
            WHEN f_score >= 3 AND m_score < 3 THEN 'Frequent low value'
            WHEN f_score < 3 AND m_score >= 3 THEN 'Big but rare'
            ELSE 'At risk'
        END AS seg
    FROM scored
)
SELECT seg,
       COUNT(*) AS subs,
       ROUND(SUM(actual_arpu_6m), 0) AS monthly_rev,
       ROUND(100.0 * SUM(actual_arpu_6m) /
             NULLIF((SELECT SUM(actual_arpu_6m) FROM customer_features), 0), 1) AS pct
FROM segments
GROUP BY seg
ORDER BY seg;
""")
```

## K7 — Revenue and churn in the At risk segment

How much monthly revenue sits in the At risk segment?

```python
q("""
WITH scored AS (
    SELECT *,
           NTILE(4) OVER (ORDER BY recharge_count) AS f_score,
           NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score
    FROM customer_features
)
SELECT COUNT(*) AS subs,
       ROUND(SUM(actual_arpu_6m), 0) AS monthly_rev,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM scored
WHERE f_score < 3 AND m_score < 3;
""")
```

## K8 — Circles with the largest Best segment

Which five circles have the highest share of Best-segment subscribers?

```python
q("""
WITH scored AS (
    SELECT *,
           NTILE(4) OVER (ORDER BY recharge_count) AS f_score,
           NTILE(4) OVER (ORDER BY recharge_total_6m) AS m_score
    FROM customer_features
)
SELECT circle_name,
       COUNT(*) AS subs,
       ROUND(100.0 * AVG(CASE WHEN f_score >= 3 AND m_score >= 3 THEN 1.0 ELSE 0.0 END), 1) AS pct_best
FROM scored
GROUP BY circle_name
ORDER BY pct_best DESC
LIMIT 5;
""")
```

# AREA M — REVENUE CONCENTRATION

## M1 — Spend share of top subscribers

What share of total spend do our top 1, 10, 100 and 1,000 subscribers bring?

```python
q("""
WITH ranked AS (
    SELECT customer_id,
           recharge_total_6m AS rev,
           ROW_NUMBER() OVER (ORDER BY recharge_total_6m DESC, customer_id) AS rn,
           SUM(recharge_total_6m) OVER (
               ORDER BY recharge_total_6m DESC, customer_id
           ) AS cum
    FROM customer_features
)
SELECT rn,
       rev,
       cum,
       ROUND(100.0 * cum /
             NULLIF((SELECT SUM(recharge_total_6m) FROM customer_features), 0), 2) AS pct
FROM ranked
WHERE rn IN (1, 10, 100, 1000)
ORDER BY rn;
""")
```

## M2 — Spend by subscriber decile

What share of total spend does each tenth of the base bring?

```python
q("""
WITH d AS (
    SELECT recharge_total_6m AS rev,
           NTILE(10) OVER (ORDER BY recharge_total_6m DESC) AS dec_
    FROM customer_features
),
g AS (
    SELECT dec_, SUM(rev) AS rev
    FROM d
    GROUP BY dec_
)
SELECT dec_,
       rev,
       ROUND(100.0 * rev / NULLIF((SELECT SUM(rev) FROM g), 0), 2) AS pct,
       ROUND(100.0 * SUM(rev) OVER (ORDER BY dec_) /
             NULLIF((SELECT SUM(rev) FROM g), 0), 2) AS cum_pct
FROM g
ORDER BY dec_;
""")
```

## M3 — Subscriber share needed to reach 80% of spend

What share of subscribers brings in 80% of total spend?

```python
q("""
WITH ranked AS (
    SELECT customer_id,
           recharge_total_6m AS rev,
           ROW_NUMBER() OVER (ORDER BY recharge_total_6m DESC, customer_id) AS rn,
           SUM(recharge_total_6m) OVER (
               ORDER BY recharge_total_6m DESC, customer_id
           ) AS cum
    FROM customer_features
),
totals AS (
    SELECT SUM(recharge_total_6m) AS total_spend,
           COUNT(*) AS total_subs
    FROM customer_features
),
needed AS (
    SELECT MIN(rn) AS subs_needed
    FROM ranked CROSS JOIN totals
    WHERE cum >= 0.8 * total_spend
)
SELECT subs_needed,
       ROUND(100.0 * subs_needed / total_subs, 1) AS pct_of_base
FROM needed CROSS JOIN totals;
""")
```

## M4 — Circles most dependent on top spenders

Which five circles depend most on their top tenth of spenders?

```python
q("""
WITH d AS (
    SELECT circle_name,
           recharge_total_6m AS rev,
           NTILE(10) OVER (
               PARTITION BY circle_name
               ORDER BY recharge_total_6m DESC
           ) AS dec_
    FROM customer_features
)
SELECT circle_name,
       ROUND(100.0 * SUM(CASE WHEN dec_ = 1 THEN rev ELSE 0 END) /
             NULLIF(SUM(rev), 0), 1) AS top_decile_pct
FROM d
GROUP BY circle_name
ORDER BY top_decile_pct DESC
LIMIT 5;
""")
```

## M5 — Revenue share from subscribers who churned

What share of monthly revenue came from subscribers who have since left?

```python
q("""
SELECT is_churned,
       COUNT(*) AS subs,
       ROUND(SUM(actual_arpu_6m), 0) AS monthly_rev,
       ROUND(100.0 * SUM(actual_arpu_6m) /
             NULLIF((SELECT SUM(actual_arpu_6m) FROM customer_features), 0), 2) AS pct
FROM customer_features
GROUP BY is_churned;
""")
```

## M6 — Concentration of data use versus spend

Is data use concentrated in the top tenth as much as spend is?

```python
q("""
WITH d AS (
    SELECT total_data_mb AS mb,
           recharge_total_6m AS rev,
           NTILE(10) OVER (ORDER BY total_data_mb DESC) AS data_dec,
           NTILE(10) OVER (ORDER BY recharge_total_6m DESC) AS spend_dec
    FROM customer_features
)
SELECT
    ROUND(100.0 * SUM(CASE WHEN data_dec = 1 THEN mb ELSE 0 END) /
          NULLIF(SUM(mb), 0), 2) AS top_decile_of_data,
    ROUND(100.0 * SUM(CASE WHEN spend_dec = 1 THEN rev ELSE 0 END) /
          NULLIF(SUM(rev), 0), 2) AS top_decile_of_revenue
FROM d;
""")
```

# AREA N — A/B EXPERIMENT

## N1 — Subscribers in each A/B group

How many subscribers are in group A and in group B?

```python
q("""
SELECT ab_group,
       COUNT(*) AS subs,
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS pct
FROM customer_features
GROUP BY ab_group;
""")
```

## N2 — Baseline similarity of A/B groups

Did groups A and B look alike before the offer began?

```python
q("""
SELECT ab_group,
       ROUND(AVG(stated_arpu), 2) AS stated_arpu,
       ROUND(100.0 * AVG(connection_type = 'Postpaid'), 2) AS pct_post,
       ROUND(AVG(tenure_days), 1) AS tenure,
       ROUND(100.0 * AVG(device_type = 'Smartphone'), 2) AS pct_smart,
       ROUND(AVG(complaint_count), 3) AS complaints
FROM customer_features
GROUP BY ab_group;
""")
```

## N3 — Monthly daily data use by A/B group

How did daily data use change, month by month, in each group?

```python
q("""
SELECT strftime('%Y-%m', u.usage_date) AS m,
       c.ab_group,
       ROUND(AVG(u.data_mb), 2) AS avg_mb
FROM usage_daily u
JOIN customer_features c
  ON c.customer_id = u.customer_id
GROUP BY m, c.ab_group
ORDER BY m, c.ab_group;
""")
```

## N4 — Data use per subscriber before and during offer

How much more data per day did a group B subscriber use during the offer?

```python
q("""
WITH per_subscriber AS (
    SELECT c.ab_group,
           u.customer_id,
           CASE
               WHEN u.usage_date >= '2025-05-01' THEN 'during'
               ELSE 'before'
           END AS period,
           AVG(u.data_mb) AS avg_mb
    FROM usage_daily u
    JOIN customer_features c
      ON c.customer_id = u.customer_id
    WHERE u.usage_date < '2025-07-01'
    GROUP BY c.ab_group, u.customer_id, period
)
SELECT period,
       ab_group,
       COUNT(*) AS subs,
       ROUND(AVG(avg_mb), 2) AS avg_mb_per_sub
FROM per_subscriber
GROUP BY period, ab_group
ORDER BY period, ab_group;
""")
```

## N5 — Churn by A/B group

What is the churn rate in group A and in group B?

```python
q("""
SELECT ab_group,
       COUNT(*) AS subs,
       SUM(is_churned) AS ch,
       ROUND(100.0 * AVG(is_churned), 2) AS churn
FROM customer_features
GROUP BY ab_group;
""")
```

## N6 — Stated and actual ARPU by A/B group

Does revenue per subscriber differ between the groups, on either ARPU column?

```python
q("""
SELECT ab_group,
       ROUND(AVG(stated_arpu), 2) AS stated,
       ROUND(AVG(actual_arpu_6m), 2) AS actual
FROM customer_features
GROUP BY ab_group;
""")
```

## N7 — Amount paid during offer months

How much did a subscriber in each group pay during the two offer months?

```python
q("""
SELECT c.ab_group,
       COUNT(*) AS subs,
       ROUND(AVG(COALESCE(w.paid, 0)), 2) AS avg_paid_may_jun
FROM customer_features c
LEFT JOIN (
    SELECT customer_id,
           SUM(amount_inr) AS paid
    FROM recharges
    WHERE recharge_date >= '2025-05-01'
      AND recharge_date < '2025-07-01'
    GROUP BY customer_id
) w
  ON w.customer_id = c.customer_id
GROUP BY c.ab_group;
""")
```

## N8 — Data allowance used during offer

What share of their data allowance did each group use during the offer?

```python
q("""
WITH d AS (
    SELECT c.ab_group,
           c.customer_id,
           c.monthly_gb,
           SUM(u.data_mb) / 1024.0 / 2.0 AS gb_per_month
    FROM customer_features c
    JOIN usage_daily u
      ON u.customer_id = c.customer_id
    WHERE u.usage_date >= '2025-05-01'
      AND u.usage_date < '2025-07-01'
    GROUP BY c.ab_group, c.customer_id, c.monthly_gb
)
SELECT ab_group,
       COUNT(*) AS subs,
       ROUND(AVG(gb_per_month), 2) AS used_gb,
       ROUND(AVG(monthly_gb), 1) AS allowance_gb,
       ROUND(100.0 * AVG(gb_per_month / NULLIF(monthly_gb, 0)), 1) AS pct_used
FROM d
GROUP BY ab_group;
""")
```
